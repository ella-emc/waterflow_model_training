# Cleaning of consolidated consumption dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
df_consumption = pd.read_csv("../../dataset/clean/training/production/tabular/consumption_consolidated.csv")
df_consumption.head()
```

### Checking for null values in each column

```python
df_consumption.isnull().sum()
```

### Checking Control Number duplicates

```python
# Check for duplicates
df_consumption.duplicated(subset=["Control Number"]).sum()

# Display duplicate rows
df_consumption[df_consumption.duplicated(subset=["Control Number"], keep=False)]
```

```python
# Keep first instance of duplicate rows
df_consumption = df_consumption.drop_duplicates(subset=["Control Number"], keep="first")

# Overwrite source dataset 
df_consumption.to_csv("../../dataset/clean/training/production/tabular/consumption_consolidated.csv", index=False)
```

```python
# Recheck for duplicate control numbers
df_consumption.duplicated(subset=["Control Number"]).sum()
```

### Checking missing districts

```python
# Display records with missing district but non-missing street
df_missing_dist_only = df_consumption[df_consumption["District"].isnull() & df_consumption["Street"].notnull()]
df_missing_dist_only.head()
```

```python
# Display records with missing district and street
df_missing_both = df_consumption[df_consumption["District"].isnull() & df_consumption["Street"].isnull()]
df_missing_both.head()
```

```python
# Save DFs to csv to misc folder
df_missing_dist_only.to_csv("../../dataset/clean/training/production/misc/missing_district_only.csv", index=False)
df_missing_both.to_csv("../../dataset/clean/training/production/misc/missing_district_street.csv", index=False)
```

### Checking missing connection type

```python
df_missing_conn_type = df_consumption[df_consumption["Connection Type"].isnull()]
df_missing_conn_type.to_csv("../../dataset/clean/training/production/misc/missing_connection_type.csv", index=False)
```

```python
df_missing_conn_type.head()
```

### Dropping insignificant rows

We drop the insignificant rows, which are determined so because they are either irrecoverable or inconsequential.

```python
year_month_cols = df_consumption.columns[5:]
year_month_cols
```

```python
# Count rows with year_month_cols all null or zero
null_or_zero_count = ((df_consumption[year_month_cols].isnull() | (df_consumption[year_month_cols] == 0)).all(axis=1)).sum()
print(f"Number of rows with all null or zero values in year_month_cols: {null_or_zero_count}")

# Drop rows with year_month_cols all null or zero
df_consumption = df_consumption[~(df_consumption[year_month_cols].isnull() | (df_consumption[year_month_cols] == 0)).all(axis=1)]

# Recheck count of rows with year_month_cols all null or zero
null_or_zero_count_after = ((df_consumption[year_month_cols].isnull() | (df_consumption[year_month_cols] == 0)).all(axis=1)).sum()
print(f"Number of rows with all null or zero values in year_month_cols after dropping: {null_or_zero_count_after}")

# Write new dataset to csv
df_consumption.to_csv("../../dataset/clean/training/production/tabular/consumption_consolidated_cleaned.csv", index=False)
```

### Export yearly readings

```python
years = ["2020", "2022", "2023", "2024", "2025"]

# Export yearly readings
for year in years:
    base_cols = ["Control Number", "District", "Street", "Connection Type"]
    
    # Get columns for the current year
    year_cols = [col for col in df_consumption.columns if col.startswith(year)]
    
    # Include 2019_DEC for 2020
    if year == "2020":
        year_cols = ["2019_DEC"] + year_cols

    columns_to_export = base_cols + year_cols

    df_year = df_consumption[columns_to_export]
    df_year.to_csv(f"../../dataset/clean/training/production/yearly/consumption_{year}_cleaned.csv", index=False)
    print(f"Exported {year} data with shape: {df_year.shape}")
    print(f"  Columns: {year_cols}")
```
